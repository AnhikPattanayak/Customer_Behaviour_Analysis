## Task 3 : Making a copy of our dataset

```python
# making an copy of our dataset

temp = df.copy()
```

## Task 4 : Filling the null values

```python
# filling the "Review Rating" null values with the median value of their respective Category

temp['Review Rating'] = temp.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
temp.isnull().sum()
```

```python
# snakecasing the column names

temp.columns = temp.columns.str.lower()
temp.columns = temp.columns.str.replace(' ','_')
temp = temp.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
temp.columns
```

## Task 5 : EDA

```python
# create a column: age_group

labels = ['Young Adult', 'Adult', 'Middle_ages', 'Senior']
temp['age_group'] = pd.qcut(temp['age'], q=4, labels = labels)
```

```python
temp[['age', 'age_group']].head(10)
```

```python
# create a column: purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-weekly': 14,
    'Annually': 365,
    'Every 3 months': 90
}

temp['purchase_frequency_days'] = temp['frequency_of_purchases'].map(frequency_mapping)
```

```python
temp[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
temp[['discount_applied', 'promo_code_used']].head(10)
```

```python
(temp['discount_applied'] == temp['promo_code_used']).all()
```

```python
# remove "promo_code_used" column

temp = temp.drop('promo_code_used', axis = 1)
```
